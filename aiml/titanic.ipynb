IQR (Interquartile Range)

IQR=Q3−Q1



```python
q1, q3 = np.percentile(df['Age'].dropna(),[25,75])
iqr = q3 -q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

clean_df = df[(df['Age'] >=lower) & (df['Age']<=upper)]
```

```python
def remove_outliers_group(df,col,group_col):
    def iqr_filter(group):
        q1 = group[col].quantile(0.25)
        q3 = group[col].quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        return group[(group[col] >= lower) & (group[col] <= upper)]
    
    return df.groupby(group_col).apply(iqr_filter).reset_index(drop=True)

clean_df = remove_outliers_group(df, 'Age','Pclass')
```

```python

plt.figure(figsize=(12,5))
sns.boxplot(x='Pclass', y='Age',data=clean_df)
plt.title("After Removing Outliers - Age")
plt.show()

```
